# Notebook 4 — Visualisasi Hasil

Membuat grafik perbandingan inference time, throughput, dan skor kualitas.

## Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.path.append('../scripts')
from utils import FIGURES_DIR
plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.size'] = 11
print('Siap visualisasi')
```

## 1. Grafik inference time (reproduksi Fig 2 & 3 jurnal)

```python
models = ['GPT-2','LLaMA-2-7B','Qwen1.5-1.8B']
fp16_70 = [23.21, 1244.29, 12.66]
int8_70 = [17.37, 18.63, 11.78]

x = np.arange(len(models))
w = 0.35
fig, ax = plt.subplots(figsize=(9,5))
bars1 = ax.bar(x-w/2, fp16_70, w, label='FP16', color='#B5D4F4', edgecolor='#185FA5', linewidth=0.5)
bars2 = ax.bar(x+w/2, int8_70, w, label='INT8', color='#5F5E5A', edgecolor='#2C2C2A', linewidth=0.5)
ax.set_xticks(x); ax.set_xticklabels(models)
ax.set_ylabel('Waktu Inferensi (detik)')
ax.set_title('Perbandingan Waktu Inferensi — RTX4070 Laptop GPU')
ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR/'inference_time_rtx4070.png', bbox_inches='tight')
plt.show()
```

## 2. Grafik skor BLEU dan ROUGE

```python
labels = ['LLaMA RTX4070','LLaMA RTX4080','Qwen RTX4070','Qwen RTX4080']
bleu   = [0.180, 0.117, 0.134, 0.113]
rouge1 = [0.509, 0.522, 0.618, 0.387]
rougeL = [0.343, 0.409, 0.291, 0.294]

x = np.arange(len(labels))
w = 0.25
fig, ax = plt.subplots(figsize=(10,5))
ax.bar(x-w,   bleu,   w, label='BLEU',    color='#F5C4B3', edgecolor='#993C1D', linewidth=0.5)
ax.bar(x,     rouge1, w, label='ROUGE-1', color='#9FE1CB', edgecolor='#0F6E56', linewidth=0.5)
ax.bar(x+w,   rougeL, w, label='ROUGE-L', color='#CECBF6', edgecolor='#534AB7', linewidth=0.5)
ax.set_xticks(x); ax.set_xticklabels(labels, rotation=15, ha='right')
ax.set_ylabel('Skor'); ax.set_ylim(0, 0.75)
ax.set_title('Skor Kualitas INT8 vs FP16 (mean, 50 prompt)')
ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR/'quality_metrics.png', bbox_inches='tight')
plt.show()
```

## 3. Scatter: Speedup vs Kualitas (trade-off plot)

```python
speedup = [1.34, 66.8, 1.88, 1.07, 1.23]
bleu_sc = [0.11, 0.18, 0.12, 0.13, 0.11]
labels2 = ['GPT2-70','LLaMA-70','LLaMA-80','Qwen-70','Qwen-80']
colors  = ['#378ADD','#D85A30','#D85A30','#1D9E75','#1D9E75']

fig, ax = plt.subplots(figsize=(8,5))
for i,(x,y,lbl,c) in enumerate(zip(speedup,bleu_sc,labels2,colors)):
    ax.scatter(x, y, s=120, color=c, zorder=3)
    ax.annotate(lbl, (x,y), textcoords='offset points', xytext=(6,4), fontsize=9)
ax.set_xlabel('Speedup INT8 vs FP16')
ax.set_ylabel('BLEU Score (INT8)')
ax.set_title('Trade-off: Kecepatan vs Kualitas per Konfigurasi')
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR/'tradeoff_scatter.png', bbox_inches='tight')
plt.show()
print('Semua grafik disimpan di outputs/figures/')
```
